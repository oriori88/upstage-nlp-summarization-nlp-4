```python
# train 데이터와 test 데이터를 불러옵니다.
DATA_PATH = "../data/"
train_data = pd.read_csv(DATA_PATH + "train.csv")
dev_data = pd.read_csv(DATA_PATH + "dev.csv")
test_data = pd.read_csv(DATA_PATH + "test.csv")

print(f"train data size: {len(train_data)}")
print(f"dev data size: {len(dev_data)}")
print(f"test data size: {len(test_data)}")
```

```python
# train data의 상위 20개 topic 분포 시각화 (bar chart)
plt.figure(figsize=(12,6))
sns.countplot(data=train_data, x='topic', order=train_data['topic'].value_counts().index)
plt.title('Train Data Topic Distribution')
plt.xlabel('Topic')
plt.ylabel('Count')
plt.xticks(rotation=45)
plt.show()
```

```python
# train data의 topic 분류
topic_counts = train_data['topic'].value_counts()
print("Train Data Topic Distribution:")
for topic, count in topic_counts.items():
    print(f"{topic}: {count}")
```

```python
# train_data의 dialogue 컬럼에서 #으로 쌓인 단어들만 추출
def extract_keywords(dialogue):
    return re.findall(r'#(\w+)', dialogue)

train_data['keywords'] = train_data['dialogue'].apply(extract_keywords)

# 추출된 키워드의 빈도수 계산
all_keywords = [keyword for keywords in train_data['keywords'] for keyword in keywords]
keyword_series = pd.Series(all_keywords)
keyword_counts = keyword_series.value_counts()#.head(20)

keyword_list = keyword_counts.index.tolist()
keyword_freq = keyword_counts.values.tolist()

print(keyword_list)
```
