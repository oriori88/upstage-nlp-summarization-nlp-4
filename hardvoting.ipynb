# 하드보팅

```python
import pandas as pd
from collections import Counter

# -----------------------------
# 하드보팅 함수
# -----------------------------
def hard_vote(preds):
    """
    preds: [summary1, summary2, summary3]
    
    규칙:
    1) 동일 문장이 2번 이상 → 그 문장 선택
    2) 아니면 가장 짧은 문장 선택
    """
    counter = Counter(preds)
    most_common, count = counter.most_common(1)[0]

    # 동일 문장이 다수결되면 그대로 사용
    if count >= 2:
        return most_common

    # fallback: 가장 짧은 문장
    return min(preds, key=len)


# -----------------------------
# submission 파일 3개 로드
# -----------------------------
sub1 = pd.read_csv('../test/4363.csv')
sub2 = pd.read_csv('../test/4340.csv')
sub3 = pd.read_csv('../test/3608.csv')

# fname 기준 정렬 (순서 뒤섞여 있을 경우 대비)
sub1 = sub1.sort_values("fname").reset_index(drop=True)
sub2 = sub2.sort_values("fname").reset_index(drop=True)
sub3 = sub3.sort_values("fname").reset_index(drop=True)

# -----------------------------
# 하드보팅 적용
# -----------------------------
final_fname = []
final_summary = []

for i in range(len(sub1)):
    fname = sub1.loc[i, "fname"]
    preds = [
        sub1.loc[i, "summary"],
        sub2.loc[i, "summary"],
        sub3.loc[i, "summary"]
    ]
    voted = hard_vote(preds)
    final_fname.append(fname)
    final_summary.append(voted)

# -----------------------------
# 최종 파일 생성
# -----------------------------
final_df = pd.DataFrame({
    "fname": final_fname,
    "summary": final_summary
})

final_df.to_csv("../test/voting/submission_hardvote_gpt.csv", index=False)

print("✔ 하드보팅 완료: submission_hardvote_gpt.csv 생성됨")
```

```python
import pandas as pd

# 1. 3개의 submission 파일 불러오기
# 실제 파일명에 맞춰서 경로를 수정해주세요.
df1 = pd.read_csv('../test/4363.csv')
df2 = pd.read_csv('../test/4340.csv')
df3 = pd.read_csv('../test/3608.csv')

print(len(df1), len(df2), len(df3))  # 세 파일의 행 개수가 동일한지 확인

# 2. 데이터프레임 생성 및 예측값 병합
target_col = 'summary'  # 예측값이 들어있는 컬럼명

voting_df = pd.DataFrame()
voting_df['model1'] = df1[target_col]
voting_df['model2'] = df2[target_col]
voting_df['model3'] = df3[target_col]

# 3. 하드 보팅 (최빈값 구하기) 수행
# axis=1은 행(row) 방향으로 계산, [0]은 최빈값이 여러 개일 때 첫 번째 값을 가져옴
final_preds = voting_df.mode(axis=1)[0]

# 4. 제출 파일 양식 만들기
submission = pd.DataFrame()
submission['fname'] = df1['fname'] # ID 컬럼 가져오기
submission[target_col] = final_preds # 하드보팅 결과 넣기

# 5. 결과 저장
submission.to_csv('../test/voting/hard_voting_submission_gemini.csv', index=False)

print("하드 보팅 완료! 'hard_voting_submission_gemini.csv' 파일이 생성되었습니다.")
print("\n--- 결과 미리보기 ---")
print(submission.head())
```
